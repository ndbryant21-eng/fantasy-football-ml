```python
import streamlit as st
import pandas as pd
import plotly.express as px

# Built-in dataset: iris (flowers)
df = px.data.iris()

# Title
st.title("🌸 Iris Flower Dashboard")

# Sidebar filter
species = st.sidebar.selectbox("Choose Species", df["species"].unique())

# Filter data
filtered_df = df[df["species"] == species]

# Show data
st.subheader(f"Data for {species}")
st.dataframe(filtered_df)

# Plot scatter
fig = px.scatter(filtered_df, x="sepal_width", y="sepal_length",
                 color="species", size="petal_length",
                 title=f"Sepal Dimensions for {species}")
st.plotly_chart(fig)

# Show summary stats
st.subheader("Summary Statistics")
st.write(filtered_df.describe())
```
