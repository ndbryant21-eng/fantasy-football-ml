```python
import pandas as pd
import time
import re
```

```python
# Seasons to scrape
seasons = list(range(2010, 2025))  # 2010 through 2024

# Base URL pattern
base_url = "https://www.pro-football-reference.com/years/{}/fantasy.htm"

# List to hold all DataFrames
fantasy_data = []

for season in seasons:
    print(f"Scraping {season} data...")
    url = base_url.format(season)
    try:
        tables = pd.read_html(url)
        df = tables[0]
        df['Season'] = season  # Add season info
        fantasy_data.append(df)
        time.sleep(2)  # Be polite to the server
    except Exception as e:
        print(f"Failed to retrieve data for {season}: {e}")

# Combine all data
all_fantasy_df = pd.concat(fantasy_data, ignore_index=True)

# Preview
print(all_fantasy_df.head())
print(f"\nData collected for {len(seasons)} seasons with {len(all_fantasy_df)} total rows.")
```

```python
clean_df = pd.read_csv('clean_fantasy_data.csv')
```

```python
def clean_player_name(name):
    # Remove trailing non-letter characters (like *, +, etc.) from the end of the string
    return re.sub(r'[\W_]+$', '', name)
```

```python
clean_df['Player'] = clean_df['Player'].astype(str).apply(clean_player_name)
```

```python
clean_df
```

```python
clean_df['Season'] = clean_df['Season'].astype(int)
```

```python
clean_df = clean_df.sort_values(['Player', 'Season'])
```

```python
clean_df
```

```python
clean_df.columns.to_list()
```

```python
features_to_lag = [
    'Age', 
    'Games_G', 
    'Games_GS',
    'Passing_Cmp', 
    'Passing_Att', 
    'Passing_Yds', 
    'Passing_TD', 
    'Passing_Int',
    'Rushing_Att', 
    'Rushing_Yds', 
    'Rushing_TD', 
    'Rushing_Y/A',
    'Receiving_Tgt', 
    'Receiving_Rec', 
    'Receiving_Yds', 
    'Receiving_TD', 
    'Receiving_Y/R',
    'Fumbles_Fmb', 
    'Fumbles_FL',
    'Scoring_TD', 
    'Scoring_2PM', 
    'Scoring_2PP',
    'Tm', 
    'FantPos',  # these will be encoded later
    'Fantasy_FantPt'
]
```

```python
# Shift each feature by 1 season per player
for feature in features_to_lag:
    clean_df[f'{feature}_prev1'] = clean_df.groupby('Player')[feature].shift(1)
```

```python
clean_df
```

```python
# Only use rows where the next season's fantasy points are known
model_df = clean_df.dropna(subset=['Fantasy_FantPt_prev1'])
```
